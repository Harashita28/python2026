Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.



```python
# taking num1 as input from user
num1 = int(input("Enter num1: "))
print("num1",num1)

# taking num2 as input from user
num2 = int(input("Enter num2: "))
print("num2",num2)

# creating a temporary variable
temp = None

# storing num1 in temp
temp = num1

# storing num2 in num1
num1 = num2

# storing original num1 (from temp) in num2
num2 = temp

# printing new values
print("After swapping:")
print("num1 =", num1)
print("num2 =", num2)
```

Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.
For example: Input: heads -> 4 legs -> 12
Output: dogs -> 2 chicken -> 2

```python
# Taking total heads and legs from the user
heads = int(input("Enter total number of heads: "))
legs = int(input("Enter total number of legs: "))

# Calculating number of dogs
dogs = (legs - (heads * 2)) // 2

# Calculating number of chickens
chickens = heads - dogs

# Printing the result
print("Number of dogs:", dogs)
print("Number of chickens:", chickens)
```

Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.


```python
n = int(input("Enter n: "))

sum = 0

for i in range(1, n + 1):
    sum = sum + i ** 2

print("Sum of squares =", sum)
```
