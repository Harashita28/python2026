### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output: 

```bash
120

````python
n = int(input("Enter a number: "))

fact = 1

for i in range(n, 0, -1):
    print(fact * i)
    fact = fact * i

print("Factorial =", fact)
````

### `Problem 10`: Write a program, which will find all such numbers between 1000 and 3000 (both included) such that each digit of the number is an even number. The numbers obtained should be printed in a space-separated sequence on a single line.

````python
for i in range(1000, 3001):

    temp = i
    valid = True

    while temp > 0:
        digit = temp % 10

        if digit % 2 != 0:
            valid = False
            break

        temp = temp // 10

    if not valid:
        continue

    print(i, end=" ")
````

###`Problem 11`: A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps. 
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps. 

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.** 

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3

````python
import math

x = 0
y = 0

print("""
Enter movement in this format:
UP 5
DOWN 3
LEFT 3
RIGHT 2

Enter ! to stop.
""")

while True:

    movement = input("Enter movement: ")

    if movement == "!":
        break

    direction, steps = movement.split()
    steps = int(steps)

    if direction == "UP":
        y = y + steps

    elif direction == "DOWN":
        y = y - steps

    elif direction == "LEFT":
        x = x - steps

    elif direction == "RIGHT":
        x = x + steps

    else:
        print("Wrong direction. Please enter UP, DOWN, LEFT or RIGHT.")
        continue

    print("Current position:", (x, y))


distance = math.sqrt(x**2 + y**2)

print("Final position:", (x, y))
print("Distance from original point:", round(distance))
````

### `Problem 12`:Write a program to print whether a given number is a prime number or not

````python
number = int(input("Enter a number: "))

prime = True

for i in range(2, number):

    if number % i == 0:
        prime = False
        break

if prime:
    print(number, "is a prime number")
else:
    print(number, "is not a prime number")
````

### `Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.


````python
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

for n in range(start, end + 1):

    # Count digits
    temp = n
    count = 0

    while temp > 0:
        count += 1
        temp = temp // 10

    # Calculate Armstrong sum
    temp = n
    total = 0

    while temp > 0:
        digit = temp % 10
        total = total + digit ** count
        temp = temp // 10

    # Check
    if total == n:
        print(n, end=" ")
````

### `Problem 14`:Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python
import math

H = int(input("Enter hour: "))
M = int(input("Enter minute: "))

hour_angle = (H % 12) * 30 + M * 0.5
minute_angle = M * 6

angle = abs(hour_angle - minute_angle)

minimum_angle = min(angle, 360 - angle)

print(math.floor(minimum_angle))
````
